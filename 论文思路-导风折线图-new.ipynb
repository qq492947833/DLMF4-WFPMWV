```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# =========================================================
# 1) 数据
# =========================================================
hours = np.array([1, 2, 3, 4, 5, 6])

model_names = [
    "Origin",
    "Prior",
    "Only",
    "P_E",
    "Lr_0.000001",
    "6_times",
    "CDF",
    "UNet",
    "CNN-Transformer",
    "Transformer",
]

rmspe_data = [
    rmspe_a_i,
    rmspe_a_i_xianyan,
    rmspe_a_i_only,
    rmspe_a_i_PE,
    rmspe_a_i_lr0000001,
    rmspe_a_i_6time,
    rmspe_a_i_cdf,
    rmspe_a_i_unet,
    rmspe_a_i_cnn_transformer,
    rmspe_a_i_transformer,
]

r_data = [
    pearsonr_a_i,
    pearsonr_a_i_xianyan,
    pearsonr_a_i_only,
    pearsonr_a_i_PE,
    pearsonr_a_i_lr0000001,
    pearsonr_a_i_6time,
    pearsonr_a_i_cdf,
    pearsonr_a_i_unet,
    pearsonr_a_i_cnn_transformer,
    pearsonr_a_i_transformer,
]

# =========================================================
# 2) 你给的真实显著性结果
#    顺序严格对应：
#    Prior, Only, P_E, Lr_0.000001, 6_times, CDF, UNet, CNN-Transformer, Transformer
# =========================================================
sig_rmspe = [
    (True, True),
    (True, False),
    (True, True),
    (True, False),
    (True, False),
    (False, False),
    (True, False),
    (True, False),
    (False, False),
]

sig_r = [
    (True, True),
    (True, False),
    (True, True),
    (False, False),
    (True, True),
    (True, False),
    (True, False),
    (True, True),
    (False, False),
]

non_origin_names = model_names[1:]
sig_map_rmspe = dict(zip(non_origin_names, sig_rmspe))
sig_map_r = dict(zip(non_origin_names, sig_r))

# 调试打印：先确认映射没错
print("RMSPE mapping:")
for k in non_origin_names:
    print(k, sig_map_rmspe[k])

print("\nR mapping:")
for k in non_origin_names:
    print(k, sig_map_r[k])

# =========================================================
# 3) 可调参数（你自己调这里）
# =========================================================
FIG_W = 9.2
FIG_H = 7.6

LEFT = 0.09
RIGHT = 0.80
TOP = 0.96
BOTTOM = 0.09

PANEL_HSPACE = 0.10

PANEL_TITLE_X = 0.00
PANEL_TITLE_Y_A = 1.005
PANEL_TITLE_Y_B = 1.005

MODELS_LEGEND_X = 0.82
MODELS_LEGEND_Y = 0.67

STYLE_LEGEND_X = 0.82
STYLE_LEGEND_Y = 0.31

RMSPE_YMIN = 0.07
RMSPE_YMAX = 0.11

R_YMIN = 0.45
R_YMAX = 0.65
R_REF_LINE = 0.50

LW_ORIGIN = 2.2
LW_OTHER = 1.8
ALPHA_ORIGIN = 1.00
ALPHA_OTHER = 0.92

ORIGIN_MARKER = "o"
ORIGIN_MARKER_SIZE = 4.5

# =========================================================
# 4) 线型映射
#    Origin / TT / 单True / FF 四类
# =========================================================
def linestyle_from_status(model_name, flag_pair):
    if model_name == "Origin":
        return "-"
    a, b = flag_pair
    if a and b:
        return "-."                 # Both significant
    elif a or b:
        return (0, (6, 2))         # One significant
    else:
        return (0, (1.2, 2.0))     # Not significant

# =========================================================
# 5) 配色
# =========================================================
model_colors = {
    "Origin": "black",
    "Prior": "#D55E00",
    "Only": "#E69F00",
    "P_E": "#009E73",
    "Lr_0.000001": "#56B4E9",
    "6_times": "#0072B2",
    "CDF": "#CC79A7",
    "UNet": "#8C564B",
    "CNN-Transformer": "#7F7F7F",
    "Transformer": "#F15BB5",
}

plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 11,
    "axes.linewidth": 1.0,
    "axes.labelsize": 12,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "legend.fontsize": 10,
})

# =========================================================
# 6) 建图
# =========================================================
fig = plt.figure(figsize=(FIG_W, FIG_H))
gs = fig.add_gridspec(
    2, 1,
    hspace=PANEL_HSPACE,
    left=LEFT,
    right=RIGHT,
    top=TOP,
    bottom=BOTTOM
)

ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[1, 0], sharex=ax1)

# =========================================================
# 7) (a) RMSPE
# =========================================================
for name, y in zip(model_names, rmspe_data):
    flags = None if name == "Origin" else sig_map_rmspe[name]

    ax1.plot(
        hours, y,
        color=model_colors[name],
        linestyle=linestyle_from_status(name, flags),
        linewidth=LW_ORIGIN if name == "Origin" else LW_OTHER,
        alpha=ALPHA_ORIGIN if name == "Origin" else ALPHA_OTHER,
        marker=ORIGIN_MARKER if name == "Origin" else None,
        markersize=ORIGIN_MARKER_SIZE if name == "Origin" else 0,
        markerfacecolor=model_colors[name] if name == "Origin" else None,
        markeredgewidth=0.0,
        zorder=5 if name == "Origin" else 3,
        label=name
    )

ax1.set_ylabel("RMSPE")
ax1.set_xlim(1, 6)
ax1.set_ylim(RMSPE_YMIN, RMSPE_YMAX)
ax1.set_xticks(hours)
ax1.grid(axis="y", linestyle="--", linewidth=0.7, alpha=0.22)
ax1.spines["top"].set_visible(False)
ax1.spines["right"].set_visible(False)
ax1.tick_params(axis="x", labelbottom=False)

ax1.text(
    PANEL_TITLE_X, PANEL_TITLE_Y_A, "(a) RMSPE",
    transform=ax1.transAxes,
    ha="left", va="bottom",
    fontsize=13, fontweight="bold"
)

# =========================================================
# 8) (b) R
# =========================================================
for name, y in zip(model_names, r_data):
    flags = None if name == "Origin" else sig_map_r[name]

    ax2.plot(
        hours, y,
        color=model_colors[name],
        linestyle=linestyle_from_status(name, flags),
        linewidth=LW_ORIGIN if name == "Origin" else LW_OTHER,
        alpha=ALPHA_ORIGIN if name == "Origin" else ALPHA_OTHER,
        marker=ORIGIN_MARKER if name == "Origin" else None,
        markersize=ORIGIN_MARKER_SIZE if name == "Origin" else 0,
        markerfacecolor=model_colors[name] if name == "Origin" else None,
        markeredgewidth=0.0,
        zorder=5 if name == "Origin" else 3,
        label=name
    )

ax2.axhline(R_REF_LINE, color="0.65", linestyle=":", linewidth=1.0)
ax2.set_ylabel("R")
ax2.set_xlabel("Hour")
ax2.set_xlim(1, 6)
ax2.set_ylim(R_YMIN, R_YMAX)
ax2.set_xticks(hours)
ax2.grid(axis="y", linestyle="--", linewidth=0.7, alpha=0.22)
ax2.spines["top"].set_visible(False)
ax2.spines["right"].set_visible(False)

ax2.text(
    PANEL_TITLE_X, PANEL_TITLE_Y_B, "(b) R",
    transform=ax2.transAxes,
    ha="left", va="bottom",
    fontsize=13, fontweight="bold"
)

# =========================================================
# 9) 右侧图例：Models
# =========================================================
model_handles = [
    Line2D(
        [0], [0],
        color=model_colors[name],
        linestyle="-",
        linewidth=2.0,
        marker="o" if name == "Origin" else None,
        markersize=4.8,
        label=name
    )
    for name in model_names
]

fig.legend(
    handles=model_handles,
    loc="center left",
    bbox_to_anchor=(MODELS_LEGEND_X, MODELS_LEGEND_Y),
    frameon=False,
    title="Models",
    title_fontsize=11,
    borderaxespad=0.0
)

# =========================================================
# 10) 右侧图例：Line style
# =========================================================
style_handles = [
    Line2D([0], [0], color="black", linestyle="-", linewidth=2.2, label="Origin"),
    Line2D([0], [0], color="black", linestyle="-.", linewidth=1.9, label="Both significant"),
    Line2D([0], [0], color="black", linestyle=(0, (6, 2)), linewidth=1.9, label="One significant"),
    Line2D([0], [0], color="black", linestyle=(0, (1.2, 2.0)), linewidth=1.9, label="Not significant"),
]

fig.legend(
    handles=style_handles,
    loc="center left",
    bbox_to_anchor=(STYLE_LEGEND_X, STYLE_LEGEND_Y),
    frameon=False,
    title="Line style",
    title_fontsize=11,
    borderaxespad=0.0
)

plt.show()
```
