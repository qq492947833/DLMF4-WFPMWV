```python
import numpy as np
import xarray as xr
import os

# ===== 输入 =====
# shap_values: (time, feature, latx, lonx, n)
# shap_values = ...
# lat = ...
# lon = ...

save_path = r"E:/huawei/result/CNN_Transformer_3_5year_order_fusion_shap.nc"

# ===== 转为 numpy 并检查 =====
shap_values = np.asarray(shap_values)

if shap_values.ndim != 5:
    raise ValueError(
        f"shap_values 应该是5维 (time, feature, latx, lonx, n)，"
        f"但当前是 {shap_values.ndim} 维，shape={shap_values.shape}"
    )

nt, nf, nlatx, nlonx, nn = shap_values.shape
print("输入 shap_values shape =", shap_values.shape)

# ===== 坐标处理 =====
time_coord = np.arange(nt)
feature_coord = np.arange(1, nf + 1)
n_coord = np.arange(1, nn + 1)

# ---- 处理 lat ----
if 'lat' in globals() and lat is not None:
    latx_coord = np.asarray(lat)
    if latx_coord.ndim != 1:
        latx_coord = np.squeeze(latx_coord)
    if latx_coord.ndim != 1:
        raise ValueError(f"lat 需要是一维坐标，当前 shape={latx_coord.shape}")
    if len(latx_coord) != nlatx:
        raise ValueError(f"lat 长度 {len(latx_coord)} != shap_values 的 latx 维度 {nlatx}")
else:
    latx_coord = np.arange(nlatx)

# ---- 处理 lon ----
if 'lon' in globals() and lon is not None:
    lonx_coord = np.asarray(lon)
    if lonx_coord.ndim != 1:
        lonx_coord = np.squeeze(lonx_coord)
    if lonx_coord.ndim != 1:
        raise ValueError(f"lon 需要是一维坐标，当前 shape={lonx_coord.shape}")
    if len(lonx_coord) != nlonx:
        raise ValueError(f"lon 长度 {len(lonx_coord)} != shap_values 的 lonx 维度 {nlonx}")
else:
    lonx_coord = np.arange(nlonx)

# ===== 构建 xarray DataArray =====
da = xr.DataArray(
    shap_values.astype(np.float32),
    dims=("time", "feature", "latx", "lonx", "n"),
    coords={
        "time": time_coord,
        "feature": feature_coord,
        "latx": latx_coord,
        "lonx": lonx_coord,
        "n": n_coord,
    },
    name="shap_values",
    attrs={
        "long_name": "SHAP values",
        "description": "SHAP values aggregated over timestep dimension; dimensions are (time, feature, latx, lonx, n)"
    }
)

ds = xr.Dataset({"shap_values": da})

# ===== 保存 =====
save_dir = os.path.dirname(save_path)
if save_dir != "":
    os.makedirs(save_dir, exist_ok=True)

encoding = {
    "shap_values": {
        "zlib": True,
        "complevel": 4,
        "dtype": "float32"
    }
}

ds.to_netcdf(save_path, encoding=encoding)

print("保存完成：")
print(save_path)
print("最终保存变量 shape =", ds["shap_values"].shape)
print("最终保存变量 dims  =", ds["shap_values"].dims)
```
